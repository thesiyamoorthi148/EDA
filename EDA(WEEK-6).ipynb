```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_healthcare_dataset.csv")
```

```python
df.head()
```

```python
sns.boxplot(
    x ="Medical Condition",
    y ="Age",
    data =df
)

plt.title("Age Distribution by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Age")
plt.xticks(rotation=45)
plt.show()
```

```python
gender_condition = pd.crosstab(
    df["Medical Condition"],
    df["Gender"]
)
gender_condition.plot(
    kind="bar",
    figsize=(10,6)
)
plt.title("Gender Distribution by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.legend(title="Gender")
plt.show()
```

```python
billing_condition = df.groupby(
    "Medical Condition"
)["Billing Amount"].mean()

billing_condition.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Average Billing Amount by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Billing Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="Insurance Provider",
    y="Billing Amount",
    data=df
)

plt.title("Billing Amount by Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.violinplot(
    x="Medical Condition",
    y="Billing Amount",
    data=df
)

plt.title("Billing Amount Distribution by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.violinplot(
    x="Insurance Provider",
    y="Billing Amount",
    data=df
)

plt.title("Billing Amount Distribution by Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x="Admission Type",
    y="Billing Amount",
    data=df
)

plt.title("Billing Amount by Admission Type")
plt.xlabel("Admission Type")
plt.ylabel("Billing Amount")

plt.show()
```

```python
monthly_admissions = df["Admission Month"].value_counts().sort_index()

print(monthly_admissions)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_admissions.index,
    monthly_admissions.values
)

plt.title("Monthly Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(range(1, 13))
plt.show()
```

```python
correlation_data =df[
    ["Age","Stay Days","Billing Amount"]
    ]
correlation_matrix = correlation_data.corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    x="Admission Type",
    data=df
)

plt.title("Distribution of Admission Types")
plt.xlabel("Admission Type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    x="Age Group",
    hue="Medical Condition",
    data=df
)

plt.title("Age Group Distribution by Medical Condition")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.show()
```
